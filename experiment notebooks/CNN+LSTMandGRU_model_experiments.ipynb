## Trying a bidirectional GRU

```python
!python -m emg2qwerty.train \
    model=conv_rnn_ctc \
    module.rnn_layers=2 \
    module.dropout=0.1 \
    module.rnn_type=gru \
    module.rnn_bidirectional=True \
    user=single_user \
    trainer.accelerator=gpu trainer.devices=1 \
    trainer.max_epochs=60
```

Better results, maybe not as good as the biLSTM, but have to run that again. 18.3 val CER, 18.9 test CER. maybe some overfitting.
